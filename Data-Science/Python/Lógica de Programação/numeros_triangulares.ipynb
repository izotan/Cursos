```python
a = [3]
n_triangular = 0
n_valor = 0
ultimo_triangular = 3

def numero_triangular(num):
  m = 1
  while num > m * (m + 1) / 2 :
    m += 1
  return num == (m * (m + 1) / 2) # Todo número triangular possui a forma (m * (m + 1) / 2), onde 'm' é um número natural maior ou igual a 1

while n_triangular <= 11: # Substituir '11' por n-1 para obter o enésimo número triangular
  if (numero_triangular(a[n_valor])): # Verdadeiro se é triangular
    n_triangular += 1 # Calcula próximo número quando atual é triangular
    a.append(a[n_valor]+1)
    ultimo_triangular = a[n_valor]
    #print(a) # DEBUG: retorna lista atual da sequência de números obtidos
    #print(f"O {n_triangular}º triangular é {ultimo_triangular}") # DEBUG: retorna individualmente o número triangular encontrado em cada loop
  else:
    a.append(2 * a[n_valor] - a[n_valor-1] + 1) # Calcula próximo número quando atual não é triangular
    ##print(a) # DEBUG: retorna lista atual da sequência de números obtidos
  n_valor += 1 # loop

print(f"O {n_triangular}º triangular é {ultimo_triangular}")
```
